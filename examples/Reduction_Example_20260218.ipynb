## Loading Data

```python
filename_pattern = "{sample}measure1_{sdd_cm}m_{energy_ev}keV_num{num}_{sample_phi_deg}deg_bpm{bpm}_id{run_id}_combined.tif"

dataset = cd_load.LoadDataset(
    var_dataset_name,                                         # the dataset name is required
    directory_path = var_directory_path,   # directory where the data is located
    # filenames=filenames,                                  # optional, list of filenames to load if you want to filter the data directory
    metadata_pattern=filename_pattern,                      # optional, filename pattern to extract metadata from the filenames
    metadata_scales={'sdd_cm': var_sdd_cm, 'energy_ev': var_energy_ev},     # optional, scales these metadata if they were in the filename with incorrect units
    data_name_pattern=var_data_name_pattern,                   # optional, pattern to follow for the data keys (each detector image in the dataset, will need to be unique), default is the filename
    metadata={'exposure_time_s': var_exposure_times_s},                        # optional, any extra metadata that applies to all data
    user_params={},                                         # optional, any unique user parameters that applies to all data
    verbose=var_verbose,                                           # optional, will provide a status bar while files are loaded
    filetype=var_export_filetype,                                         # optional, specify the filetype, this will help filter the files if they have not been filtered already, if not provided the laoder will try to guess based on the file extension
    detector_type=var_detector_type,                                # optional, will try to extract count time and pixel size from the pilatus header, if this isn't possible it will assume 172 um as the pixel size
)
```

### Find sample detector distance and phi offset

```python

phis = []
sdds = []

for data in dataset.datas.values():
    if abs(data.metadata["sample_phi_deg"]) <= 5:
        sdd, sdd_err, fig = data.find_sdd_from_reference_peaks(100, width_qdy_px=sdd_phi_width_qdy_px, width_qdx_px=sdd_phi_width_qdx_px, beam_center_guess=sdd_phi_beam_center_guess, min_distance=sdd_phi_min_dist, threshold_abs=sdd_phi_thresh_abs, exclude_q=[(sdd_phi_exclude_q_low, sdd_phi_exclude_q_high)], num_peaks=sdd_phi_num_peaks, peak_orders=sdd_phi_peak_orders)
        phis.append(data.metadata["sample_phi_deg"])
        sdds.append(sdd)

        if EXPORT_PLOTS:
            plt.savefig(f"./export/{timestamp}/plots/plot_sdd_phi{data.metadata['sample_phi_deg']}_{timestamp}.png", dpi=300)
```

```python
z = np.polyfit(phis, sdds, 2)

plt.figure()
plt.scatter(phis, sdds)
fit_phis = np.arange(-5.5, 5.5, 0.1)
plt.plot(fit_phis, np.poly1d(z)(fit_phis), color='red')

min_phi = -z[1]/(2*z[0])
sample_phi_offset_deg = -min_phi # this value is added to the nominal phi value to determine actual phi value, i.e., actual_phi = nominal_phi + sample_phi_offset_deg
plt.vlines(min_phi, min(sdds), max(sdds), color='black')
sdd_fit = np.poly1d(z)(min_phi)

if EXPORT_PLOTS:
    plt.savefig(f"./export/{timestamp}/plots/plot_sdd_fitcurve_{timestamp}.png", dpi=300)
```

```python
dataset.update_all_metadata(dict(sample_phi_offset_deg=sample_phi_offset_deg, sdd_cm=sdd_fit))
```

### Find Omega Correction

```python
data_subset = [ #pick keys/conditions that make sense for your data
                #also select ones that you are confident have good, clear peaks
    "-45.0",
    "-35.0",
    "-30.0",
    "-20.0",
    "20.0",
    "30.0",
    "35.0",
    "45.0",
]
omegas = {}

for data_key in data_subset:
    if data_key in dataset.datas.keys():
        print(f"Finding Omega for {data_key}")
        temp_data = dataset.datas[data_key]
        omega, _ = temp_data.find_omega_from_peaks( 
            width_qdy_px=omega_corr_width_qdy_px, # or you can set an index range!
            width_qdx_px=omega_corr_width_qdx_px,
            log_scale=omega_corr_log_scale,
            refinement_size=omega_corr_refinement_size, # box around each peak sent to the Gaussian refinement function
            show_plot=omega_corr_show_plot, 
            zoom_plot=omega_corr_zoom_plot, # if set to True, the search box with a small border will be shown in the plot to more easily see the peaks found, otherwise if set to False they will all be overlaid on the original detector image
            plotting_kwargs={},
            exclude_q = (omega_corr_exclude_q_low, omega_corr_exclude_q_high),
            min_distance=omega_corr_min_dist, # see documentation for scikit-image peak_local_max() for all keyword arguments you can pass through
            threshold_rel=omega_corr_thresh_rel, # if log_scale is set to True, the data is sent to the peak finding algorithm after taking log10 so make sure these values make sense for data after the transformation
        )
        omegas[data_key] = omega
        print(f"Omega angle found for {data_key}: {omega} degrees")

        if EXPORT_PLOTS:
            plt.savefig(f"./export/{timestamp}/plots/plot_omega_{data_key}_{timestamp}.png", dpi=300)
```

```python
sum = 0
for key, value in omegas.items():
    sum += value
average_omega = sum / len(omegas)
print(f"Average omega correction to apply: {average_omega} degrees")
dataset.update_all_metadata({"sample_omega_deg": average_omega})
```
